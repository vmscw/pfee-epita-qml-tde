# Les métriques dans le Machine Learning tabulaire

Ce document présente les principales métriques utilisées en machine learning sur données tabulaires. Elles sont regroupées selon leur nature : classification, régression, probabilistes, calibration et complexité. Ces métriques sont largement utilisées dans le monde réel pour mesurer l'efficacité d'un algorithme selon un mode d'utilisation.

---

## I. Métrique de classification

Les métriques de classification interviennent lorsque la variable cible est discrète, c’est-à-dire qu’elle prend un ensemble fini de valeurs (classes). Elles permettent d’évaluer la capacité d’un modèle à correctement prédire une étiquette ou à estimer une probabilité associée à chaque classe.

Dans les problèmes tabulaires, ces métriques sont particulièrement importantes car les jeux de données présentent souvent des déséquilibres de classes, des distributions hétérogènes ou des coûts d’erreur asymétriques.

### I.1 Accuracy

L’accuracy mesure la proportion de prédictions correctes sur l’ensemble des observations.

$$
\text{Accuracy} = \frac{1}{N} \sum_{i=1}^{N} \mathbb{1}(\hat{y}_i = y_i)
$$
Elle est simple à interpréter mais peut être trompeuse lorsque les classes sont déséquilibrées, car un modèle naïf peut obtenir une accuracy élevée en prédisant systématiquement la classe majoritaire.

### I.2 Precision

La precision mesure de vrais positifs parmi les prédictions positives.

$$\text{Precision} = \frac{TP}{TP + FP}$$

avec

$TP = \text{nombres de bonnes prédictions}$

et 

$FP = \text{nombre de mauvaises prédictions}$

Tout aussi simple à interpréter et qui complète au sein d'une même classe l'accuracy (voir I.1).
Elle est particulièrement importante lorsque les faux positifs sont coûteux

### I.3 Recall

Le recall mesure la capacité du modèle à détecter les positifs réels.

$$\text{Recall} = \frac{TP}{TP + FN}$$

Le recall permet de chiffrer le taux d'erreur au sein d'une même classe et de sa bonne représentation par l'algorithme, il permet de voir si certaines classes sont bien représentées ou au contraire mal représentées par l'algorithme en fonction de certains critères.
Il est crucial lorsque les faux négatifs sont critiques

### I.4 F1-score

Le F1-score combine précision et rappel en une seule métrique.

$$F_1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$

Il est particulièrement utile lorsque les classes sont déséquilibrées et que l’on souhaite un compromis entre les deux types d’erreurs.

### I.5 Confusion Matrix (structure)

\begin{bmatrix} TP & FP \\ FN & TN \end{bmatrix}

Structure de base pour dériver plusieurs métriques.

## II. Métriques probabilistes

### II.1 Log Loss (Cross-Entropy)

$$\text{LogLoss} = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right]$$

Mesure la qualité des probabilités prédites.

### II.2 ROC-AUC

$$\text{ROC-AUC} = P( s(x^+) > s(x^-) )$$

Probabilité qu’un positif soit mieux scoré qu’un négatif.

### II.3 PR-AUC (Precision-Recall AUC)

$$\text{PR-AUC} = \int_0^1 \text{Precision}(r) , d\text{Recall}(r)$$

Plus informative que ROC-AUC en cas de fort déséquilibre.

---

## III. Métriques de régression

### III.1 MAE

$$\text{MAE} = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i|$$

Erreur moyenne absolue.

### III.2 MSE

$$\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$

Pénalise fortement les grandes erreurs.

### III.3 RMSE

$$\text{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2}$$

Version interprétable du MSE.

### III.4 R² score

$$R^2 = 1 - \frac{\sum_{i=1}^{N} (y_i - \hat{y}i)^2}{\sum{i=1}^{N} (y_i - \bar{y})^2}$$

Proportion de variance expliquée.

---

## IV. Métriques de calibration

### IV.1 Brier Score

$$\text{Brier} = \frac{1}{N} \sum_{i=1}^{N} (p_i - y_i)^2$$

Mesure la qualité des probabilités prédites.

### IV.2 Expected Calibration Error (ECE)

$$\text{ECE} = \sum_{m=1}^{M} \frac{|B_m|}{N} \left| \text{acc}(B_m) - \text{conf}(B_m) \right|$$

Mesure l’écart entre confiance et précision réelle.

---

## V. Métriques de robustesse et généralisation

### V.1 Cross-validation score

$$\text{CV} = \frac{1}{K} \sum_{k=1}^{K} s_k$$

Moyenne des performances sur K folds.

### V.2 Variance inter-fold

$$\text{Var}{CV} = \frac{1}{K} \sum{k=1}^{K} (s_k - \bar{s})^2$$

Mesure la stabilité du modèle.

---

## VI. Métriques de complexité

### VI.1 Training time

$$T_{train} = \sum_{b=1}^{B} t_b$$

### VI.2 Inference time

$$T_{infer} = \frac{1}{N} \sum_{i=1}^{N} t(x_i)$$

### VI.3 Memory footprint

$$M_{peak} = \max_t M(t)$$

---

## VII. Synthèse

En machine learning tabulaire, aucune métrique unique ne suffit à évaluer un modèle. L’évaluation moderne repose sur une combinaison de :

- performance prédictive (Accuracy, AUC, RMSE)
- qualité probabiliste (LogLoss, Brier, ECE)
- robustesse statistique (CV, variance)
- coût computationnel (train/inference time)
- stabilité et généralisation

Cette approche multi-critères est essentielle pour comparer équitablement des familles de modèles très différentes, comme les gradient boosting, les réseaux de neurones et les modèles foundation.
