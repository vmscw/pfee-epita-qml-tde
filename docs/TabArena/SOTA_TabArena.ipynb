# State of the Art - Tab Arena

Ce document présente les différentes métriques utilisées pour mesurer la puissance d'un modèle, les résultats des différents modèles tabulaires ainsi que leur classement en fin mai 2026.

Les différents modèles tabulaires utilisés lors de cette "compétition" proviennent majoritairement de bibliothèques.

---------

## I. Généralités

TabArena est un benchmark de référence dédié à l’évaluation des modèles de machine learning sur données tabulaires. Son objectif principal est de fournir un cadre standardisé, reproductible et évolutif permettant de comparer de manière fiable différentes approches, allant des modèles classiques de gradient boosting aux modèles récents de type foundation models.

Contrairement à des benchmarks statiques, TabArena adopte une approche “living benchmark”, c’est-à-dire un système mis à jour continuellement avec de nouveaux datasets, de nouveaux modèles et des protocoles d’évaluation ajustés. Cette dynamique permet de limiter l’obsolescence rapide des résultats observée dans de nombreux benchmarks académiques traditionnels.

### I.1 Principe de reproductibilité et conditions de comparaison

Un des objectifs centraux de TabArena est de garantir une reproductibilité stricte des expériences. Cela implique que chaque modèle est évalué dans des conditions standardisées, permettant une comparaison équitable.

Les principaux éléments de standardisation sont les suivants :

- Splits de données identiques : tous les modèles utilisent les mêmes partitions train / validation / test.
- Seeds fixes : les initialisations aléatoires sont contrôlées afin de réduire la variance expérimentale.
- Préprocessing homogène : les transformations appliquées aux données (encodage, normalisation, gestion des valeurs manquantes) suivent des règles communes.
- Budget de calcul contrôlé : les modèles sont évalués sous des contraintes similaires en termes de temps d’entraînement et de ressources.
- Protocole d’évaluation uniforme : les mêmes métriques et procédures de test sont appliquées à tous les modèles.

L’objectif de ces contraintes est d’éviter les biais classiques de comparaison (tuning asymétrique, preprocessing spécifique à un modèle, sur-optimisation locale) et de garantir une lecture fiable des performances relatives.

### I.2. Cas d’utilisation des tâches tabulaires

TabArena couvre plusieurs types de problèmes supervisés sur données tabulaires, qui correspondent aux cas d’usage les plus fréquents en machine learning appliqué.

#### Classification binaire

Il s’agit de prédire une variable cible prenant deux valeurs possibles (ex : churn client, fraude, défaut de paiement).

#### Classification multiclasse

La cible peut prendre plus de deux classes distinctes (ex : type de produit, catégorie d’image structurée, diagnostic médical multi-classes).

#### Régression

L’objectif est de prédire une variable continue (ex : prix immobilier, durée, score de risque, demande future).

#### Variantes et contraintes fréquentes

Selon les datasets, certaines difficultés supplémentaires peuvent apparaître :

- forte cardinalité des variables catégorielles
- déséquilibre des classes
- données manquantes
- relations non linéaires complexes
- bruit important dans les features

Ces différents cas permettent d’évaluer la robustesse des modèles dans des conditions proches de la réalité.

### I.3. Métriques utilisées pour l’évaluation

L’évaluation des modèles dans TabArena repose sur des métriques adaptées à chaque type de tâche, afin de mesurer la qualité des prédictions de manière cohérente.

#### Classification

Les métriques les plus courantes sont :

- Accuracy : proportion de prédictions correctes
- ROC-AUC : capacité à séparer les classes positives et négatives
- F1-score : compromis entre précision et rappel, utile en cas de déséquilibre

#### Régression

Les métriques principales sont :

- RMSE (Root Mean Squared Error) : pénalise fortement les grandes erreurs
- MAE (Mean Absolute Error) : mesure robuste des écarts moyens
- R² score : proportion de variance expliquée par le modèle
Agrégation et comparaison globale

Au-delà des métriques individuelles, TabArena utilise des mécanismes d’agrégation permettant de comparer les modèles sur l’ensemble des datasets. Cela peut inclure :

- moyennes normalisées des scores
- rankings relatifs
- systèmes de type Elo basés sur des confrontations par pair

Ces approches permettent de produire un classement global plus stable que de simples moyennes de performance.

## II. Diversité des modèles pour données tabulaires

Les données tabulaires constituent un cadre central du machine learning appliqué. Contrairement à d’autres domaines comme la vision ou le langage, il n’existe pas une architecture dominante unique, mais plusieurs familles de modèles aux inductive biases différents. Cette section présente les principales catégories de modèles utilisées dans TabArena et dans l’état de l’art actuel.

### II.1 Liste des différents modèles

Les modèles évalués sur données tabulaires peuvent être regroupés en quatre grandes familles :

- Modèles basés sur les arbres de décision et le gradient boosting
  - Random Forest
  - XGBoost
  - LightGBM
  - CatBoost
  - EBM
  - ExtraTrees
  - Perpetual Booster
  - RandomForest
- Modèles de deep learning spécialisés pour données tabulaires
  - TabNet
  - FT-Transformer
  - SAINT
  - MLP améliorés (ResMLP tabulaire)
  - FastaiMLP
  - TabM
  - ModernNCA
  - RealMLP
  - TorchMLP
- Modèles foundation pour données tabulaires
  - TabPFN (et variantes comme TabPFN-2.x ou RealTabPFN-2.x)
  - TabDPT
  - SAP-RPT-OSS
  - TabICLv.x
  - TabStar
- Frameworks AutoML et ensembles de modèles
  - AutoGluon Tabular
  - H2O AutoML
  - solutions d’ensembling multi-modèles

### II.2 Modèles basés sur les arbres et le boosting

Les modèles de type arbre constituent historiquement la référence sur les données tabulaires. Leur succès repose sur leur capacité à capturer efficacement des relations non linéaires et des interactions entre variables sans nécessiter un prétraitement complexe.

Les algorithmes de gradient boosting tels que XGBoost, LightGBM et CatBoost construisent des ensembles d’arbres séquentiellement, où chaque nouvel arbre corrige les erreurs des précédents.

#### Utilisation principale :

- datasets de taille moyenne à grande
- problèmes industriels structurés
- forte présence de variables catégorielles et numériques

#### Avantages :

- excellente performance empirique
- robustesse au bruit et aux données hétérogènes
- faible besoin en feature engineering avancé

#### Limites :

- moins efficaces sur très petits datasets sans tuning
- parallélisation parfois limitée (selon implémentation)
- difficulté à capturer certaines structures globales complexes

### II.3 Modèles de deep learning pour données tabulaires

Les modèles de deep learning tabulaire cherchent à reproduire le succès des réseaux de neurones dans d’autres domaines en adaptant leurs architectures aux spécificités des données structurées.

On retrouve notamment :

- TabNet, basé sur un mécanisme d’attention et de sélection de features
- FT-Transformer, qui applique l’architecture Transformer aux colonnes tabulaires
- SAINT, qui introduit des mécanismes d’attention entre lignes et colonnes

#### Utilisation principale :

- datasets riches en interactions complexes
- grands volumes de données
- recherche académique et expérimentation

#### Avantages :

- capacité à modéliser des interactions non linéaires complexes
- flexibilité architecturale
- potentiel de généralisation important avec beaucoup de données

#### Limites :

- nécessite souvent plus de données que les GBDT
- tuning plus délicat
- performance parfois instable selon les datasets

#### II.4 Modèles foundation pour données tabulaires

Les modèles foundation tabulaires constituent une approche récente visant à généraliser l’apprentissage sur données structurées. Ils sont pré-entraînés sur un grand nombre de tâches synthétiques ou réelles afin d’acquérir une forme de “prior” sur la structure des données tabulaires.

TabPFN est l’un des représentants les plus connus de cette famille. Il utilise une approche d’in-context learning pour inférer directement des prédictions à partir d’un petit ensemble d’exemples sans entraînement classique.

#### Utilisation principale :

- petits à moyens datasets
- prototypage rapide
- situations avec peu de tuning possible

#### Avantages :

- très faible besoin de configuration
- performances compétitives sur petits datasets
- inference rapide sans entraînement lourd

#### Limites :

- scalabilité limitée
- dépendance aux distributions vues en pré-entraînement
- moins interprétable que les modèles classiques

### II.5 AutoML et ensembles de modèles

Les approches AutoML visent à automatiser l’ensemble du pipeline de machine learning, incluant la sélection de modèles, le tuning d’hyperparamètres et l’ensembling.

Des frameworks comme AutoGluon ou H2O AutoML combinent plusieurs modèles (souvent des GBDT, réseaux de neurones et modèles linéaires) pour produire une solution robuste.

#### Utilisation principale :

- production industrielle
- compétition de machine learning
- scénarios nécessitant des performances stables sans expertise forte

#### Avantages :

- très bonne robustesse globale
- exploite les forces de plusieurs modèles
- réduit le besoin de tuning manuel

#### Limites :

- coût computationnel élevé
- manque de transparence sur la décision finale
- complexité d’interprétation
